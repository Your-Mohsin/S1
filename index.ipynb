## Selecting the most useful columns from the dataset

```python
useful_cols = [
    "Area",
    "Location",
    "No. of Bedrooms",
    "City",
    "CarParking",
    "SwimmingPool",
    "24X7Security",
    "LiftAvailable",
    "ClubHouse",
    "Price"
]

df = df[useful_cols]
```

## Model Building

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor # Other regression models
import joblib
```

### Encoding string values to numberic

```python
city_encoder = LabelEncoder()
df["City"] = city_encoder.fit_transform(df["City"])
location_encoder = LabelEncoder()
df["Location"] = location_encoder.fit_transform(df["Location"])
```

### Saving the encoders

```python
joblib.dump(city_encoder,"city_encoder.joblib") # Saving city_encoder
joblib.dump(location_encoder,"location_encoder.joblib") # Saving location_encoder
```

### Splitting X and Y

```python
X = df.drop(columns="Price")
Y = df["Price"]
```

```python
model1 = LinearRegression()
model1.fit(X,Y)
```

```python
y_pred1 = model1.predict(X)
print("R2 Score:",r2_score(Y,y_pred1))
```

```python
model2 = GradientBoostingRegressor()
model2.fit(X,Y)
```

```python
y_pred2 = model2.predict(X)
print("R2 Score: ",r2_score(Y,y_pred2))
```

### Saving the model

```python
joblib.dump(model2,"model.joblib")
```

### Saving the cleaned Data
